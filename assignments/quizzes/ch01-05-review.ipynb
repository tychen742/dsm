```python
### 13. Calculate Audit Metrics
# The lab calculated an order total. Here, calculate simple management metrics from case counts.
# Use the variables below to calculate open_cases and closure_rate.
# Print exactly:
# Open cases: 17
# Closure rate: 0.67

starting_cases = 20
new_cases = 7
closed_cases = 10
assigned_cases = 15

### Your code starts here.



### Your code ends here.
```

```python
### Solution
starting_cases = 20
new_cases = 7
closed_cases = 10
assigned_cases = 15

open_cases = starting_cases + new_cases - closed_cases
closure_rate = closed_cases / assigned_cases

print("Open cases:", open_cases)
print("Closure rate:", round(closure_rate, 2))
```

```python
### 15. Count Until Closing
# Use a while loop to add customer counts until the sentinel value 0 appears.
# Print exactly:
# Days counted: 2
# Customers before closing: 80
daily_customers = [42, 38, 0, 51]

### Your code starts here.



### Your code ends here.
```

```python
### Solution
daily_customers = [42, 38, 0, 51]

index = 0
days_counted = 0
customers_before_closing = 0
while daily_customers[index] != 0:
    customers_before_closing += daily_customers[index]
    days_counted += 1
    index += 1

print("Days counted:", days_counted)
print("Customers before closing:", customers_before_closing)
```

```python
### 16. Compare to Target
# Build a list showing whether each store met the target.
# Print exactly:
# Met target: [False, True, True, False]
sales = [80, 105, 110, 95]
target = 100

### Your code starts here.



### Your code ends here.
```

```python
### 17. Universal Function Style
# Square each demand error and total the squared errors.
# Store the squared values in squared_errors.
# Print exactly:
# Squared errors: [4, 1, 9, 16]
# Total squared error: 30
errors = [2, -1, 3, -4]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
errors = [2, -1, 3, -4]

squared_errors = []
for error in errors:
    squared_errors.append(error ** 2)

total_squared_error = sum(squared_errors)
print("Squared errors:", squared_errors)
print("Total squared error:", total_squared_error)
```

```python
### 18. Match Product Rows
# Import pandas as pd, then create a DataFrame with product and price columns.
# Filter rows where product is "pen" and print the number of matched rows and the pen price.
# Print exactly:
# Matched rows: 2
# Pen price: 2.5
products = [["pen", 2.5], ["notebook", 4.0], ["pen", 2.5]]

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import pandas as pd

products = [["pen", 2.5], ["notebook", 4.0], ["pen", 2.5]]

products_df = pd.DataFrame(products, columns=["product", "price"])
pen_rows = products_df[products_df["product"] == "pen"]
pen_price = pen_rows.iloc[0]["price"]

print("Matched rows:", len(pen_rows))
print("Pen price:", pen_price)
```

```python
### 19. Create a Customer Table
# Import pandas as pd, then create a DataFrame from names and purchases.
# Store the DataFrame in a variable named customers.
# Print exactly:
# Customer rows: 3
# Purchase total: 12
names = ["Ana", "Bo", "Cy"]
purchases = [5, 3, 4]

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import pandas as pd

names = ["Ana", "Bo", "Cy"]
purchases = [5, 3, 4]

customers = pd.DataFrame({"name": names, "purchases": purchases})

print("Customer rows:", len(customers))
print("Purchase total:", int(customers["purchases"].sum()))
```

```python
### 20. Select Labeled Values
# Import pandas as pd, then create a Series from values with labels as its index.
# Store the Series in a variable named scores.
# Print exactly:
# B score: 91
# First score: 82
values = [82, 91, 76]
labels = ["A", "B", "C"]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import pandas as pd

values = [82, 91, 76]
labels = ["A", "B", "C"]

scores = pd.Series(values, index=labels)

print("B score:", scores.loc["B"])
print("First score:", scores.iloc[0])
```
