# Estimation

This chapter shifts from testing claims to estimating unknown quantities. You will use percentiles, bootstrap resampling, and confidence intervals to describe uncertainty around estimates.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Estimation | Using sample data to approximate an unknown population value. |
| 2 | Parameter | A numerical feature of a population. |
| 3 | Statistic | A numerical summary computed from a sample. |
| 4 | Point estimate | A single statistic used as an estimate of a parameter. |
| 5 | Percentile | A value below which a stated percentage of observations fall. |
| 6 | Median | The 50th percentile. |
| 7 | Bootstrap | A resampling method that samples with replacement from observed data. |
| 8 | Resample | A new sample drawn from the original sample, usually with replacement. |
| 9 | With replacement | Sampling where selected observations can appear more than once. |
| 10 | Bootstrap distribution | The distribution of a statistic across many resamples. |
| 11 | Confidence interval | An interval estimate intended to capture plausible parameter values. |
| 12 | Confidence level | The long-run success rate associated with an interval method. |
| 13 | Margin of error | A distance added around an estimate to express uncertainty. |
| 14 | Sampling variability | Variation in statistics across possible samples. |
| 15 | Population | The full group or process of interest. |
| 16 | Sample | The observed subset used for estimation. |
| 17 | Bias | Systematic error in an estimate or sampling process. |
| 18 | Precision | How tightly an estimate is pinned down. |
| 19 | Uncertainty | The amount of doubt around an estimate. |
| 20 | Empirical distribution | A distribution formed from observed values. |
| 21 | Interval estimate | A range of plausible values rather than a single number. |
