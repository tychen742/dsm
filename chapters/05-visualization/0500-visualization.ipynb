# Visualization

This chapter introduces visualization as a management decision tool. You will learn how to choose chart types, create quick plots with pandas, compare common chart families, and polish figures so patterns are easier to read.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Data visualization | Using charts and graphs to reveal patterns in data. |
| 2 | Audience | The people who need to understand or act on a chart. |
| 3 | Chart type | A visual form such as a bar chart, line chart, or scatter plot. |
| 4 | Categorical variable | A variable whose values are labels or groups. |
| 5 | Quantitative variable | A variable measured with numeric values. |
| 6 | Frequency | The count of observations in a category or interval. |
| 7 | Relative frequency | A frequency expressed as a proportion or percentage. |
| 8 | Bar chart | A chart that compares values across categories. |
| 9 | Line chart | A chart that shows change across an ordered sequence. |
| 10 | Scatter plot | A chart that displays paired numeric values as points. |
| 11 | Histogram | A chart that shows the distribution of one numeric variable. |
| 12 | Box plot | A compact summary of median, quartiles, and possible outliers. |
| 13 | Axis | A reference line that provides the scale for values. |
| 14 | Legend | A guide that maps colors or symbols to groups. |
| 15 | Label | Text that names axes, categories, or marks. |
| 16 | Scale | The mapping between data values and visual positions. |
| 17 | Outlier | An observation that sits far from the main pattern. |
| 18 | Pandas plotting | The DataFrame and Series plotting interface built on Matplotlib. |
| 19 | Matplotlib | A flexible Python library for building and customizing plots. |
| 20 | Seaborn | A statistical visualization library built on Matplotlib. |
| 21 | Encoding | The use of position, color, size, or shape to represent data. |
