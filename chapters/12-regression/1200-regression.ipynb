
<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Regression | A method for modeling or predicting an outcome from one or more features. |
| 2 | Prediction | An estimate of an outcome for a new or existing case. |
| 3 | Feature | An input variable used to predict an outcome. |
| 4 | Target | The outcome variable being predicted. |
| 5 | Scatter plot | A plot showing paired numeric values. |
| 6 | Association | A relationship or pattern between variables. |
| 7 | Correlation | A standardized measure of linear association. |
| 8 | Regression line | A straight line used to predict an outcome from a feature. |
| 9 | Slope | The predicted change in the outcome for a one-unit change in the feature. |
| 10 | Intercept | The predicted outcome when the feature is zero. |
| 11 | Fitted value | The model prediction for an observed case. |
| 12 | Residual | The observed value minus the fitted value. |
| 13 | Least squares | A method that chooses the line with the smallest sum of squared residuals. |
| 14 | Residual plot | A plot of residuals used to check model patterns. |
| 15 | Standard units | Values expressed in standard deviations from the mean. |
| 16 | Root mean squared error | A typical prediction error measured in outcome units. |
| 17 | Linearity | The assumption that a straight line captures the main trend. |
| 18 | Extrapolation | Prediction outside the range of observed data. |
| 19 | Influential point | An observation that strongly affects a fitted model. |
| 20 | Diagnostics | Checks used to evaluate how well a model fits. |
| 21 | Overfitting | Fitting noise instead of a durable pattern. |
