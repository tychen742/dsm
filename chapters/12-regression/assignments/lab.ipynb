```python
### Solution
import numpy as np
import pandas as pd

ads = pd.DataFrame({
    "ad_spend": [2, 3, 4, 5, 6, 7, 8, 9],          # $1,000s per week
    "sales": [60, 72, 70, 85, 88, 98, 95, 110],    # $1,000s per week
})

def standard_units(x):
    return (x - np.mean(x)) / np.std(x)

r = np.mean(standard_units(ads['ad_spend']) * standard_units(ads['sales']))
print("r:", round(r, 3))
```

```python
### Solution
import numpy as np
import pandas as pd

ads = pd.DataFrame({
    "ad_spend": [2, 3, 4, 5, 6, 7, 8, 9],          # $1,000s per week
    "sales": [60, 72, 70, 85, 88, 98, 95, 110],    # $1,000s per week
})

def standard_units(x):
    return (x - np.mean(x)) / np.std(x)

def correlation(x, y):
    return np.mean(standard_units(x) * standard_units(y))

def slope(x, y):
    return correlation(x, y) * np.std(y) / np.std(x)

def intercept(x, y):
    return np.mean(y) - slope(x, y) * np.mean(x)

print("Slope:", round(slope(ads['ad_spend'], ads['sales']), 2))
print("Intercept:", round(intercept(ads['ad_spend'], ads['sales']), 2))
```

```python
### Solution
import numpy as np
import pandas as pd

ads = pd.DataFrame({
    "ad_spend": [2, 3, 4, 5, 6, 7, 8, 9],          # $1,000s per week
    "sales": [60, 72, 70, 85, 88, 98, 95, 110],    # $1,000s per week
})

def standard_units(x):
    return (x - np.mean(x)) / np.std(x)

def correlation(x, y):
    return np.mean(standard_units(x) * standard_units(y))

def slope(x, y):
    return correlation(x, y) * np.std(y) / np.std(x)

def intercept(x, y):
    return np.mean(y) - slope(x, y) * np.mean(x)

a = slope(ads['ad_spend'], ads['sales'])
b = intercept(ads['ad_spend'], ads['sales'])

print("Predicted sales at 6.5:", round(a * 6.5 + b, 2))
print("Residual at 5:", round(85 - (a * 5 + b), 2))
```

```python
### Question 4: Least Squares Beats a Rule of Thumb
# A manager uses the rule of thumb: sales = 50 + 6 * ad_spend.
#   1. Compute the root mean squared error (RMSE) of the regression line's
#      fitted values: sqrt(mean of (actual - fitted) ** 2).
#   2. Compute the RMSE of the rule of thumb's predictions.
#   3. Round both to 2 decimals.
# Print exactly:
# Regression RMSE: 3.67
# Rule of thumb RMSE: 4.27
import numpy as np
import pandas as pd

ads = pd.DataFrame({
    "ad_spend": [2, 3, 4, 5, 6, 7, 8, 9],          # $1,000s per week
    "sales": [60, 72, 70, 85, 88, 98, 95, 110],    # $1,000s per week
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

ads = pd.DataFrame({
    "ad_spend": [2, 3, 4, 5, 6, 7, 8, 9],          # $1,000s per week
    "sales": [60, 72, 70, 85, 88, 98, 95, 110],    # $1,000s per week
})

def standard_units(x):
    return (x - np.mean(x)) / np.std(x)

def correlation(x, y):
    return np.mean(standard_units(x) * standard_units(y))

def slope(x, y):
    return correlation(x, y) * np.std(y) / np.std(x)

def intercept(x, y):
    return np.mean(y) - slope(x, y) * np.mean(x)

def rmse(predicted):
    return np.sqrt(np.mean((ads['sales'] - predicted) ** 2))

fitted = slope(ads['ad_spend'], ads['sales']) * ads['ad_spend'] + intercept(ads['ad_spend'], ads['sales'])
print("Regression RMSE:", round(rmse(fitted), 2))
print("Rule of thumb RMSE:", round(rmse(50 + 6 * ads['ad_spend']), 2))
# No straight line has a smaller RMSE than the least squares line.
```

```python
### Question 5: Reading the Residuals
# A café tracked units sold of a pastry at different prices (in dollars).
#   1. Fit the regression line of units on price and compute the residuals
#      (actual minus fitted).
#   2. The rows are sorted by price. Compute the mean residual for the first
#      3 rows (low prices), the middle 3, and the last 3 (high prices),
#      each rounded to 1 decimal.
#   3. If the low and high groups are positive and the middle is negative,
#      print "curved"; otherwise print "no clear pattern".
# Print exactly:
# Low prices: 4.0
# Middle prices: -8.7
# High prices: 4.6
# Pattern: curved
import numpy as np
import pandas as pd

cafe = pd.DataFrame({
    "price": [5, 6, 7, 8, 9, 10, 11, 12, 13],
    "units": [90, 70, 55, 44, 37, 33, 31, 30, 30],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

cafe = pd.DataFrame({
    "price": [5, 6, 7, 8, 9, 10, 11, 12, 13],
    "units": [90, 70, 55, 44, 37, 33, 31, 30, 30],
})

def standard_units(x):
    return (x - np.mean(x)) / np.std(x)

def correlation(x, y):
    return np.mean(standard_units(x) * standard_units(y))

def slope(x, y):
    return correlation(x, y) * np.std(y) / np.std(x)

def intercept(x, y):
    return np.mean(y) - slope(x, y) * np.mean(x)

a = slope(cafe['price'], cafe['units'])
b = intercept(cafe['price'], cafe['units'])
residuals = cafe['units'] - (a * cafe['price'] + b)

low = round(residuals.iloc[0:3].mean(), 1)
middle = round(residuals.iloc[3:6].mean(), 1)
high = round(residuals.iloc[6:9].mean(), 1)

if low > 0 and middle < 0 and high > 0:
    pattern = "curved"
else:
    pattern = "no clear pattern"

print("Low prices:", low)
print("Middle prices:", middle)
print("High prices:", high)
print("Pattern:", pattern)
# The line is too low at both ends and too high in the middle, so a curve
# would describe the relationship better than a straight line.
```

(function () {
  var labId = 'ch12-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';

    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var cm = cell.querySelector('.CodeMirror');
    if (cm && cm.CodeMirror && typeof cm.CodeMirror.getValue === 'function') return cm.CodeMirror.getValue();
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent && cmContent.cmView && cmContent.cmView.view) return cmContent.cmView.view.state.doc.toString();
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.prototype.slice.call(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch12-lab-feedback-list');
    list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row.querySelector('span');
      var item = feedback[question];
      row.className = 'ch12-lab-feedback-row';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch12-lab-feedback-list');
    list.hidden = true;
    grader.querySelectorAll('.ch12-lab-feedback-row').forEach(function (row) {
      row.className = 'ch12-lab-feedback-row';
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var payload = text ? JSON.parse(text) : null;
        if (!payload || !response.ok || !payload.ok) {
          throw new Error((payload && (payload.message || payload.error)) || 'Submission API is unavailable.');
        }
        return payload;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch12-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var actions = grader.querySelector('.ch12-lab-actions');
    var result = document.getElementById('ch12-lab-result');
    var submit = document.getElementById('ch12-lab-submit');
    var reset = document.getElementById('ch12-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

      clearFeedback(grader);
      submit.disabled = true;
      result.textContent = 'Submitting code.';

      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;

          applyFeedback(grader, payload.feedback || {});
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
