# Python installation

Unlike Unix-like systems (macOS and Linux), Windows does not ship with a default Python. It is, however, very common for an OS to have multiple Python installations overtime because certain third-party software packages (e.g., Visual Studio IDE, Anaconda) may bundle Python as dependencies when installing.



## Check Existing Python


Since Python is widely used by other tools and packages, you more than likely have some versions of Python automatically installed in your system when installing other packages. Let us check which version(s) of Python are already installed on your computer. 

::::{tab-set}
:::{tab-item} Windows
You can use the **Python Launcher** for Windows (`py`.exe) and the command-finder (`where`.exe) to see the current Python installation(s) in Windows PowerShell.

The `where.exe` command finder shows all Python installations and their installation paths:

```powershell
PS C:\Users\[user]> where.exe python
PS C:\Users\[user]\workspace\py> where.exe python
C:\Users\[user]\AppData\Local\Programs\Python\Python312\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python313\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python310\python.exe
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\python.exe
PS C:\Users\[user]\workspace\py>
```

The Python Launcher shows the version information and the *launcher default* Python version (*):
```powershell
PS C:\Users\[user]> py -0
 -V:3.13          Python 3.13 (64-bit)
 -V:3.13-arm64    Python 3.13 (Store)
 -V:3.12          Python 3.12 (64-bit)
 -V:3.10 *        Python 3.10 (64-bit)
PS C:\Users\[user]>
```
:::

:::{tab-item} macOS
With macOS, you already have a version of Python shipped with the OS. You may issue the following commands in the system shell to see the version(s) of Python you have in the current system. (In most cases, you may need to use `python3` instead of `python`.)

Use `python3 --version` to 
```bash
[user]@[host]:~/workspace/dsm$ python3 --version
Python 3.12.2
[user]@[host]:~/workspace/dsm$ which python3
/Library/Frameworks/Python.framework/Versions/3.12/bin/python3
[user]@[host]:~/workspace/dsm$ 
```
:::
::::

## Install Python

::::{tab-set}
:::{tab-item} Windows

```{note}
If you search for or type `python` in PowerShell, Windows may bring you to Microsoft Store for Python installation. Avoid this Microsoft Store version if you want Python in PATH, full `pip` package installs, and Python Launcher (`py`) for Windows.
```

Go to python.org/downloads and click the big "Download Python 3.x" button for current release. It auto-detects the OS and gives you the latest stable release.

Or, choose a specific stable release version, go to  https://www.python.org/downloads/windows/. Find the release of your choice and you would mostly choose the Windows installer (64-bit). 

Start the installation and 
```{figure} ../../figures/python_windows_install.png
:alt: python_windows_install
:width: 400px
:name: python_windows_install
Python Installation Options in Windows
```

:::

:::{tab-item} macOS

Go to python.org/downloads and click the big "Download Python 3.x" button for current release. It auto-detects the OS and gives you the latest stable release.

Or, choose a specific stable release, go to  https://www.python.org/downloads/macos/. Find the release of your choice and download the macOS installer. 


:::
::::

## Choose Python Version

### Choose Python version temporarily

In case you have more than one Python version installed, you may need to control the version of Python that you use for your projects. If the default Python is different from the desired version, you may use the version **suffixes** to designate the desired Python version. This helps, e.g., when you need to create a virtual environment and you need to designate a Python version such as 3.12.

::::{tab-set}
:::{tab-item} Windows
You may use the Python launcher to decide which Python version to use by appending the major and minor version number to the launcher: `py -Major.Minor`. 

When you issue just `py` in PowerShell to launch the Python shell, you get the default launcher version (in this case, Python 3.10):

```powershell
PS C:\Users\tychen\workspace\py> py
Python 3.10.10 (tags/v3.10.10:aad5f6a, Feb  7 2023, 17:20:36) 
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

You may choose to run a specific version, such as 3.13, by running `py -3.13`:
```powershell
PS C:\Users\tychen\workspace\py> py -3.13
Python 3.13.7 (tags/v3.13.7:bcee1c3, Aug 14 2025, 14:15:11) 
Type "help", "copyright", "credits" or "license" for more information.
>>>
```
:::

:::{tab-item} macOS
In macOS, you append the Python version number in the format of `Major.Minor` when issuing `python` to choose the Python version to use:

```bash
$ python3.12
Python 3.12.2 (v3.12.2:6abddd9f6a, Feb  6 2024, 17:02:06) 
[Clang 13.0.0 (clang-1300.0.29.30)] on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```
:::
::::

```{figure} ../../figures/python_EV_order.png
:alt: python_EV_order
:width: 450px
:name: python_EV-order
The top Python path is the default Python
```

## Resources

- [How to Install Python on Your System: A Guide](https://realpython.com/installing-python/#windows-how-to-check-or-get-python) from realpython.com
- [How to Install Python on Windows 10](https://www.digitalocean.com/community/tutorials/install-python-windows-10)
