# Python installation

```{contents}
:local:

```


Unlike Unix-like systems (macOS and Linux), Windows does not ship with a default Python. It is, however, very common for your computer, Windows or not, to have multiple Python releases installed already. That could have happened because Python adoption has been prevalent and, overtime, the software packages (e.g., Visual Studio IDE, Anaconda, etc.) you use may have bundled Python as a dependency when you install them.


:::{admonition} Terminal & Shell

This section requires fundamental knowledge about command line interface (CLI), including how to launch the terminal application and how to operate the operating system shell by issuing commands in the system shell.

- **Terminal**: A terminal is the program you open to access the **shell**. Common terminal programs include PowerShell/Command Prompt on Windows, Terminal on macOS, and Terminal on Linux.
- **Shell**: The shell is what actually interprets and runs your commands, acting as the interface between you and the operating system. Common shells include `PowerShell`/Command Prompt (Windows), `bash` (macOS and Linux), and zsh (macOS).
- **REPL**: A REPL (Read-Eval-Print Loop) is an interactive environment for a specific programming language, not the OS. You type a line of code, it runs immediately, and the result prints right away. Examples include the **Python shell** (`>>>`at the command line), `csharprepl` for C#, and Node.js's built-in REPL. Jupyter Notebook is also built on a REPL: each notebook cell is a REPL to make each notebook a persistent, non-linear, editable document where cells can be run out of order and outputs can include plots and rich media, not just text.
:::


## Check Existing Python


Since Python is widely used by other tools and packages, you more than likely have some versions of Python automatically installed in your system when installing other packages. To check which version(s) of Python are already installed on your computer, start the terminal application. 

:::{warning}
If you do not have Python installed and you type `python` in Windows Search bar or `python` in PowerShell then enter, Windows will show you a Microsoft Store suggestion for Python installation. **DO NOT install Python from Microsoft Store.** It usually leads to issues with Python execution permission, `pip` installer, and lack of `py.exe` launcher. Instead, go to next section: [Install Python](#install-python) to learn how to download a standalone Python installer from [python.org](python.org) and install Python.
:::

### Start Terminal
Start the terminal application in your computer to use the CLI.

::::{tab-set}
:::{tab-item} Windows

Launch the terminal application (`Windows PowerShell`) so you can type commands at the system shell prompt (`>` with a flashing vertical bar `|`). 

```powershell
Windows PowerShell
Copyright (C) Microsoft Corporation. All rights reserved.

Loading personal and system profiles took 1119ms.
PS C:\Users\[user]> |
```
:::

:::{tab-item} macOS
From Spotlight Search, look for the terminal application (`Terminal.app`) and return to launch it so you can type commands at the system shell prompt (`$` or `%` followed by a block):
```bash
Last login: Fri Aug 28 12:18:57 on ttys000

[user]@[computer]:~$ █
```
:::
::::

### Check Default Python

Now you are at the command line, issue the commands to check the default Python release. 

::::{tab-set}
:::{tab-item} Windows

To check the default Python version, issue `python --version` and enter. If a Python is installed, you will see the returned Python version information.
```powershell
PS C:\Users\[user]> python --version
Python 3.13.7
```

If no Python is installed, you may see the following error message. Go to next section: [Install Python](#install-python) to learn how to download a standalone Python installer from [python.org](python.org) and install Python.
```powershell
PS C:\Users\[user]>
Python was not found; run without arguments to install from the Microsoft Store, or disable this shortcut from Settings > Apps > Advanced app settings > App execution aliases.

PS C:\Users\[user]>
```
:::

:::{tab-item} macOS
With macOS, you already have a version of Python shipped with the OS. You may issue the following commands in the system shell to see all the versions of Python you have in the current system. Note that, in most cases, you may need to use `python3` instead of `python`. 

To find the default Python version when you enter `python3` (or `python`) when you enter it at the command line, pass the `--version` option to Python.
```bash
[user]@[host]:~$ python3 --version
Python 3.12.2
```
:::
::::

### Check Python Installations

In addition to the default installation, you can check how many Python releases are installed in your computer. 

::::{tab-set}
:::{tab-item} Windows
You can use the command-finder (`where`.exe) and the **Python Launcher** for Windows (`py`.exe) to see the current Python installation(s) in Windows PowerShell. 

The `where.exe` command finder shows all Python installations and their installation paths:

```powershell
PS C:\Users\[user]> where.exe python
PS C:\Users\[user]\workspace\py> where.exe python
C:\Users\[user]\AppData\Local\Programs\Python\Python312\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python313\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python310\python.exe
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\python.exe
PS C:\Users\[user]\workspace\py>
```

The Python Launcher shows the version information and the *launcher default* Python version (the entry with `*`). You can use `py -0` or `py --list` for this.
```powershell
PS C:\Users\[user]> py -0
 -V:3.13          Python 3.13 (64-bit)
 -V:3.13-arm64    Python 3.13 (Store)
 -V:3.12          Python 3.12 (64-bit)
 -V:3.10 *        Python 3.10 (64-bit)
PS C:\Users\[user]>
```
:::

:::{tab-item} macOS

To check all the installed python versions and their installation paths, use the `which` command (note that this gives you the installations records in the `PATH` environment variable):
```bash
[user]@[host]:~$ which -a python python3
/opt/homebrew/bin/python3
/usr/local/bin/python3
/usr/bin/python3
/opt/homebrew/bin/python3
[user]@[host]:~$
```

To see the version information for the installations above, use the following code and enter.
```bash
[user]@[host]:~$
for p in $(which -a python python3); do
    echo "$p -> $($p --version 2>&1)"
done

```

You will see the output with version information as below.
```bash
[user]@[host]:~$
for p in $(which -a python python3); do
    echo "$p -> $($p --version 2>&1)"
done
/opt/homebrew/bin/python3 -> Python 3.14.5
/usr/local/bin/python3 -> Python 3.12.10
/usr/bin/python3 -> Python 3.9.6
/opt/homebrew/bin/python3 -> Python 3.14.5
[user]@[host]:~$
```

To see the Python installations through Homebrew (recommended), issue the commands: 
```bash
[user]@[host]:~$ brew list | grep python@
python@3.12
python@3.13
python@3.14
```

Actually, all your Homebrew Python installations are in `/opt/homebrew/Cellar/` with symbolic links in `/opt/homebrew/opt/`. You can `cd` into those folders and `ls -l python*` to see all of them listed. 
:::
::::

## Install Python

If there is not current Python installation on your computer, or if you would like to install a current or a specific version of Python, follow the instructions here to install it. For both Windows and macOS operating systems, multiple Python installation usually does not cause problems.

```{note}
If you have already installed Python through the **Microsoft Store**, you are advised to perform a python.org installation following the instructions here. 
```

::::{tab-set}
:::{tab-item} Windows

Visit [python.org/downloads](python.org/downloads) and you should see a big yellow button saying "Download Python install manager." **DO NOT install** this `.msix` package version. This is a community-maintained Microsoft Store installation package and may have issues with file write permissions and certain path behaviors because it's a Store app. 

Instead, below the big yellow Download button, find **Or get the standalone installer for [Python 3.major.minor](https://www.python.org/ftp/python/3.14.7/python-3.14.7-amd64.exe)**. The Python 3.major.minor is the most recent *stable* release and would look like **Python 3.14.7**. Click on **Python 3.14.7** to install the standalone installer, which is a python.org release and is a larger `.exe` file. 

If you would like to choose a specific release version, which sometimes is required by certain software packages, go to [python.org/downloads/windows/](https://python.org/downloads/windows/), find the release of your choice, and you choose the right Windows installer.

After the download is complete, click on the `python-*major*.*minor*.*micro***.exe**` (`python-3.14.7-amd64.exe` in this case) installer file to start the installation. You will see the first installation screen as below. Make sure you check both checkboxes:
  - **Use admin privileges when installing py.exe (this enables py.exe to manage all Python installations)**
  - **Add python.exe to the PATH environment variable (this sets this installation as the default Python)**

```{figure} ../../../figures/python_install_windows_3.14.7.png
:alt: python_windows_install
:width: 400px
:name: python_windows_install
Python Installation Options in Windows
```

Click on **Install Now** to start installation. Note that here you see the installation path under Install Now. It should be something like `C:\Users\[user]\AppData\Local\Programs\Python\Python314`. 

You will then answer "Yes" to continue on next window, then you will see two other installation windows: *Setup Progress* and *Setup was successful* to click on *Close* to complete the installation. 

:::

:::{tab-item} macOS
Go to [python.org/downloads](python.org/downloads). The website will auto-detect your OS and gives you the latest stable release. Click the big yellow "Download Python 3.major.minor" (something like *Download Python 3.14.7*) download button to download the Python installer. The installer file extension is `.pkg`. 

If you would like to choose a specific release version, which sometimes is required by certain software packages, go to [python.org/downloads/macos/](https://python.org/downloads/macos/), find the release of your choice, and you choose the right macOS installer.

After the download is complete, click on the `python-*major*.*minor*-macos11.pkg` (python-3.14.7-macos11 in this case) installer file to start the installation. You will see the first installation screen as below. 

```{figure} ../../../figures/python_macos_install_3.14.7.png
:alt: python_macos_install
:width: 400px
:name: python_macos_install
Python Installation Window in macOS
```
You click on *Continue* to continue the installation. You will see the windows for Read Me and License, which you both click on *Continue* to the software license agreement, which you *Agree* to the installation window, then you click on *Install* to install. After the *Running package scripts window*, you will see a *successfully installed* window for you to close the installer. 

Upon successful installation of Python, a Finder window will open to show you the location of the installation, which should be `/Applications/Python 3.14/`. 

:::
::::

## Choose Python Version

### Choose Python version temporarily

When you have more than one Python version installed, sometimes you need to control the version of Python that you use for your projects. If the default Python is different from the desired version, which happens often, you may use the Python version **suffixes** to designate the Python version to run. This helps, e.g., when you want to use a specific version of Python for your specific virtual environment.

::::{tab-set}
:::{tab-item} Windows
You may use the Python launcher to decide which Python version to use by appending the major and minor version number of the installed Python to the launcher: `py -*major*.*minor*` (or `py -V:*major*.*minor*`). 

Supposed that your Python Launcher (`py.exe`) has a default Python of 3.10. When you enter `py` to start Python, you always get the default launcher version:

```powershell
PS C:\Users\[user]\workspace\py> py
Python 3.10.10 (tags/v3.10.10:aad5f6a, Feb  7 2023, 17:20:36) 
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

You may choose to run a specific installed version, such as 3.14, by running `py -3.14` to get to the version of Python shell:
```powershell
PS C:\Users\[user]\workspace\py> py -3.14
Python 3.14.7 (tags/v3.14.7:bcee1c3, Aug 14 2026, 14:15:11) 
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

Usually, the newly installed version will be the default versions when you run `python` (first `python.exe` found in PATH variable) instead of `py` launcher. You can therefore just use `python` to use the PATH default version if that is what you want. The version of your default `python`.exe can be found by issuing `python --version`.

```powershell
PS C:\Users\[user]> python --version
Python 3.14.7
PS C:\Users\[user]> python
Python 3.14.7 (tags/v3.14.7:823f032, Aug  5 2026, 10:51:32) [MSC v.1944 64 bit (AMD64)] on win32
Type "help", "copyright", "credits" or "license" for more information.
>>>
```
:::

:::{tab-item} macOS
You may choose to decide which Python version to use by appending the major and minor version number of the installed Python to python: `python3 -*major*.*minor*`. 

Supposed that your default Python of 3.14.5. When you enter `python3` to start Python, you get the default launcher version, which you can confirm using `python3 --version`.

```bash
[user]@[host]:~$ python3 --version
Python 3.14.5
[user]@[host]:~$ 
```

```bash
[user]@[host]:~$ python3
Python 3.14.5 (main, May 10 2026, 10:21:34) [Clang 21.0.0 (clang-2100.0.123.102)] on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```

Now suppose you want to use a specific non-default installed version, such as 3.12. You issue `python3.12` to get to the version of Python shell:
```bash
[user]@[host]:~$ python3.12
Python 3.12.13 (main, Mar  3 2026, 12:39:30)
Type "help", "copyright", "credits" or "license" for more information.
>>>
```
:::
