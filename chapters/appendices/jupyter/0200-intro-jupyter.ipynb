# Jupyter Setup

Jupyter Notebook is a crucial component of the Python ecosystem and is widely used by the data science and machine learning communities for its ability to combine code, notes, analysis and visualization seamlessly in one document. Alternatively, you can use Google Colab, which is a cloud version of Jupyter Notebook. 

````{note}
Some prefer Anaconda, which is a huge bundle of tools and libraries. With some command line skills, Jupyter Notebook standalone is leaner, faster, avoids licensing concerns, mirrors real-world Python workflows, and is simply better project control. The popular code editor/IDE VS Code can also run Jupyter Notebooks, but the user experience differs.
````

| Tool             | What it is                                  | What it does    |
| ---------------- | --------------------------------------------|-----------------|
| Jupyter Notebook | An interactive notebook interface           | Interactive execution |
| VS Code          | A general-purpose code editor/IDE           | Write and manage code | 
| Anaconda         | A Python distribution + environment manager | Install and manage packages, including Jupyter |
| Google Colab     | A cloud-hosted Jupyter Notebook service     | Runs Jupyter notebooks in cloud/browser |

The best way to install and configure Jupyter Notebook is to create a Python virtual environment. To do that, you need to use the Command Line Interface (CLI) on your computer. To install and start using Jupyter Notebook, follow the four steps below: 

1. Install Python 
2. Create project directory
3. Create Python virtual environment (`.venv`) in the project folder
4. Start Jupyter Notebook
5. Create and edit notebooks
