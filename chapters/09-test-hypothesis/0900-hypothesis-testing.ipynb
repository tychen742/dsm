
<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Hypothesis test | A method for comparing observed data with a stated chance model. |
| 2 | Null hypothesis | The baseline claim tested against the data. |
| 3 | Alternative hypothesis | The competing claim suggested by the research question. |
| 4 | Test statistic | A numeric measure used to summarize evidence. |
| 5 | Observed statistic | The test statistic computed from the actual data. |
| 6 | Null distribution | The distribution of a test statistic assuming the null hypothesis is true. |
| 7 | P-value | The probability, under the null model, of evidence at least as extreme as observed. |
| 8 | Significance level | A cutoff used to decide whether a p-value is small enough to reject the null. |
| 9 | Reject the null | Conclude that the observed evidence is inconsistent with the null model. |
| 10 | Fail to reject | Decide that the data do not provide enough evidence against the null model. |
| 11 | One-sided test | A test where the alternative points in one direction. |
| 12 | Two-sided test | A test where the alternative allows differences in either direction. |
| 13 | Simulation | Repeated random trials used to approximate a distribution. |
| 14 | Random assignment | Assigning units to groups by chance. |
| 15 | Chance model | A model describing what outcomes would look like under randomness alone. |
| 16 | Statistical significance | Evidence judged unlikely under the null model. |
| 17 | Practical significance | The real-world size or importance of an effect. |
| 18 | Type I error | Rejecting a true null hypothesis. |
| 19 | Type II error | Failing to reject a false null hypothesis. |
| 20 | Inference | Using sample data to reason about a broader process. |
| 21 | Decision rule | A rule for converting statistical evidence into an action. |
