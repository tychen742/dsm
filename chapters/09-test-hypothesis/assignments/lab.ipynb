```python
### Question 1: Observed Test Statistic
# A call center's model says 20% of calls are escalated to a manager.
# In a random sample of 400 calls, 98 were escalated.
#   1. Compute the observed percent of escalated calls.
#   2. Compute the test statistic: the distance between the observed percent
#      and the model's 20%, | observed percent - 20 |, rounded to 2 decimals.
# Print exactly:
# Observed percent: 24.5
# Test statistic: 4.5
import numpy as np
import pandas as pd

escalated = 98
sample_size = 400
model_percent = 20
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

escalated = 98
sample_size = 400
model_percent = 20

observed_percent = round(100 * escalated / sample_size, 2)
test_statistic = round(abs(observed_percent - model_percent), 2)

print("Observed percent:", observed_percent)
print("Test statistic:", test_statistic)
```

```python
### Question 2: Total Variation Distance
# A loyalty program should reflect the store's customer base. The table shows the
# share of all customers in each region and the share of program members.
#   1. Compute the absolute difference between the two distributions for each region.
#   2. Compute the total variation distance (TVD): the sum of the absolute
#      differences divided by 2, rounded to 2 decimals.
# Print exactly:
# TVD: 0.1
import numpy as np
import pandas as pd

regions = pd.DataFrame({
    "region": ["North", "Central", "South"],
    "all_customers": [0.40, 0.35, 0.25],
    "members": [0.30, 0.40, 0.30],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

regions = pd.DataFrame({
    "region": ["North", "Central", "South"],
    "all_customers": [0.40, 0.35, 0.25],
    "members": [0.30, 0.40, 0.30],
})

abs_diff = (regions['members'] - regions['all_customers']).abs()
tvd = round(abs_diff.sum() / 2, 2)

print("TVD:", tvd)
```

```python
### Question 3: P-value From a Simulation
# To test whether members were picked at random from all customers, an analyst
# simulated 20 random samples under that model and computed the TVD for each.
# Large TVDs are evidence against the model.
#   1. Count the simulated TVDs that are at least as large as the observed TVD.
#   2. Compute the p-value: that count divided by the number of simulations.
# Print exactly:
# At least as large: 4
# P-value: 0.2
import numpy as np
import pandas as pd

observed_tvd = 0.10
simulated_tvds = pd.Series([0.04, 0.07, 0.01, 0.08, 0.09, 0.05, 0.07, 0.07, 0.05, 0.03,
                            0.00, 0.10, 0.06, 0.14, 0.08, 0.07, 0.04, 0.10, 0.04, 0.11])
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

observed_tvd = 0.10
simulated_tvds = pd.Series([0.04, 0.07, 0.01, 0.08, 0.09, 0.05, 0.07, 0.07, 0.05, 0.03,
                            0.00, 0.10, 0.06, 0.14, 0.08, 0.07, 0.04, 0.10, 0.04, 0.11])

at_least_as_large = (simulated_tvds >= observed_tvd).sum()
p_value = at_least_as_large / len(simulated_tvds)

print("At least as large:", at_least_as_large)
print("P-value:", p_value)
```

```python
### Question 4: Applying the Conventions
# Three tests were run on different store policies. Label each result with the
# conventions from this chapter:
#   p-value below 0.01 -> "highly significant"
#   p-value below 0.05 -> "significant"
#   otherwise          -> "not significant"
# Print one line per test, in table order.
# Print exactly:
# Free shipping: highly significant
# Loyalty email: significant
# New layout: not significant
import numpy as np
import pandas as pd

tests = pd.DataFrame({
    "policy": ["Free shipping", "Loyalty email", "New layout"],
    "p_value": [0.003, 0.04, 0.21],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

tests = pd.DataFrame({
    "policy": ["Free shipping", "Loyalty email", "New layout"],
    "p_value": [0.003, 0.04, 0.21],
})

for policy, p_value in zip(tests['policy'], tests['p_value']):
    if p_value < 0.01:
        label = "highly significant"
    elif p_value < 0.05:
        label = "significant"
    else:
        label = "not significant"
    print(policy + ":", label)
```

```python
### Question 5: False Alarms
# A team ran 20 A/A tests, comparing a web page with an identical copy of itself,
# so the null hypothesis (no difference) is true in every test.
#   1. Count the tests whose p-value is below the 0.05 cutoff. Each one is a
#      Type I error (a false alarm).
#   2. Compute the false-alarm rate: that count divided by the number of tests.
# Print exactly:
# False alarms: 1
# False-alarm rate: 0.05
import numpy as np
import pandas as pd

p_values = pd.Series([0.280, 0.783, 0.988, 0.986, 0.883, 0.913, 0.708, 0.554, 0.923, 0.031,
                      0.370, 0.606, 0.478, 0.793, 0.195, 0.062, 0.127, 0.963, 0.133, 0.226])
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

p_values = pd.Series([0.280, 0.783, 0.988, 0.986, 0.883, 0.913, 0.708, 0.554, 0.923, 0.031,
                      0.370, 0.606, 0.478, 0.793, 0.195, 0.062, 0.127, 0.963, 0.133, 0.226])

false_alarms = (p_values < 0.05).sum()

print("False alarms:", false_alarms)
print("False-alarm rate:", false_alarms / len(p_values))
# With a 5% cutoff, about 5% of tests of a true null hypothesis come out significant.
```

<script>
(function () {
  var labId = 'ch09-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';

    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var cm = cell.querySelector('.CodeMirror');
    if (cm && cm.CodeMirror && typeof cm.CodeMirror.getValue === 'function') return cm.CodeMirror.getValue();
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent && cmContent.cmView && cmContent.cmView.view) return cmContent.cmView.view.state.doc.toString();
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.prototype.slice.call(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch09-lab-feedback-list');
    list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row.querySelector('span');
      var item = feedback[question];
      row.className = 'ch09-lab-feedback-row';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch09-lab-feedback-list');
    list.hidden = true;
    grader.querySelectorAll('.ch09-lab-feedback-row').forEach(function (row) {
      row.className = 'ch09-lab-feedback-row';
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var payload = text ? JSON.parse(text) : null;
        if (!payload || !response.ok || !payload.ok) {
          throw new Error((payload && (payload.message || payload.error)) || 'Submission API is unavailable.');
        }
        return payload;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch09-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var actions = grader.querySelector('.ch09-lab-actions');
    var result = document.getElementById('ch09-lab-result');
    var submit = document.getElementById('ch09-lab-submit');
    var reset = document.getElementById('ch09-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      submit.disabled = true;
      result.textContent = 'Submitting code.';

      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          applyFeedback(grader, payload.feedback || {});
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
