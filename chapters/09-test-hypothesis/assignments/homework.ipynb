**2. Reading a p-Value**

A test of a new pricing policy gives a p-value of 0.03. This means there is a 3% chance that the null hypothesis is true.

<div class="tf-options" role="radiogroup" aria-label="Question 2 answer">
  <label><input type="radio" name="ch09-hw-q2" value="true"> True</label>
  <label><input type="radio" name="ch09-hw-q2" value="false"> False</label>
</div>

::::


```python
### Solution
# False
# The p-value is the chance, computed assuming the null hypothesis is true, of a test statistic at least as extreme as the one observed. It is not the probability that the null hypothesis is true.
```

**3. Choosing a Test Statistic**

An analyst checks whether a jury panel's mix of five ethnic groups matches the eligible population. Large values of the total variation distance between the panel and the population are evidence against the model of random selection.

<div class="tf-options" role="radiogroup" aria-label="Question 3 answer">

  <label><input type="radio" name="ch09-hw-q3" value="false"> False</label>
</div>

::::


```python
### Solution
# True
# The TVD measures how far apart the two distributions are. Random selection tends to give small TVDs, so large values favor the alternative.
```

**5. Many Tests**

A marketing team runs 20 independent tests in which the null hypothesis is actually true, using a 5% cutoff. They should expect about one of them to come out statistically significant.

::::


```python
### Solution
# True
# With a 5% cutoff, a true null hypothesis is rejected about 5% of the time, and 5% of 20 tests is 1. That significant result is a Type I error (a false alarm).
```

```python
### Solution
import numpy as np
import pandas as pd

defective = 42
shipment = 600
claimed_percent = 5

observed_percent = round(100 * defective / shipment, 2)
print("Observed percent:", observed_percent)
print("Test statistic:", round(abs(observed_percent - claimed_percent), 2))
```

```python
### 7. Channel Mix TVD
# Compare this month's sales channel mix with last year's mix.
#   1. Compute the total variation distance between the two distributions,
#      rounded to 2 decimals.
# Print exactly:
# TVD: 0.08
import numpy as np
import pandas as pd

mix = pd.DataFrame({
    "channel": ["Web", "App", "Store", "Phone"],
    "last_year": [0.40, 0.25, 0.30, 0.05],
    "this_month": [0.42, 0.31, 0.22, 0.05],
})

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

mix = pd.DataFrame({
    "channel": ["Web", "App", "Store", "Phone"],
    "last_year": [0.40, 0.25, 0.30, 0.05],
    "this_month": [0.42, 0.31, 0.22, 0.05],
})

tvd = round((mix['this_month'] - mix['last_year']).abs().sum() / 2, 2)
print("TVD:", tvd)
```

```python
### 8. P-value for the Defect Rate
# Under the supplier's 5% claim, an analyst simulated 25 shipments of 600 parts and
# computed the test statistic | simulated percent - 5 | for each one.
#   1. Compute the p-value: the proportion of simulated statistics at least as
#      large as the observed statistic of 2.0.
# Print exactly:
# P-value: 0.08
import numpy as np
import pandas as pd

observed_statistic = 2.0
simulated = pd.Series([0.0, 1.2, 0.2, 0.8, 1.1, 0.2, 0.8, 2.1, 1.3, 1.2, 0.3, 0.7, 0.3,
                       1.1, 0.9, 0.4, 1.1, 1.1, 0.1, 0.7, 0.4, 0.4, 1.6, 2.6, 0.2])

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

observed_statistic = 2.0
simulated = pd.Series([0.0, 1.2, 0.2, 0.8, 1.1, 0.2, 0.8, 2.1, 1.3, 1.2, 0.3, 0.7, 0.3,
                       1.1, 0.9, 0.4, 1.1, 1.1, 0.1, 0.7, 0.4, 0.4, 1.6, 2.6, 0.2])

p_value = (simulated >= observed_statistic).mean()
print("P-value:", p_value)
```

```python
### 9. The Cutoff Matters
# The defect-rate test gave a p-value of 0.08.
#   1. Decide at a 5% cutoff: print "reject" if the p-value is below the cutoff,
#      otherwise "do not reject".
#   2. Decide again at a 10% cutoff.
# Print exactly:
# At 0.05: do not reject
# At 0.1: reject
import numpy as np
import pandas as pd

p_value = 0.08
cutoffs = [0.05, 0.1]

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

p_value = 0.08
cutoffs = [0.05, 0.1]

for cutoff in cutoffs:
    if p_value < cutoff:
        decision = "reject"
    else:
        decision = "do not reject"
    print("At " + str(cutoff) + ":", decision)
# Report the p-value along with the decision, so readers can judge for themselves.
```

```python
### Solution
import numpy as np
import pandas as pd

observed_tvd = 0.071
simulated = pd.Series([0.037, 0.028, 0.044, 0.051, 0.065, 0.050, 0.062, 0.010, 0.031, 0.025,
                       0.054, 0.061, 0.013, 0.046, 0.068, 0.073, 0.022, 0.064, 0.058, 0.039])

cutoff_value = round(np.percentile(simulated, 95), 3)
print("95th percentile:", cutoff_value)
print("Observed is above it:", observed_tvd > cutoff_value)
```

<script>
(function () {
  var homeworkId = 'ch09-homework';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';

    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var cm = cell.querySelector('.CodeMirror');
    if (cm && cm.CodeMirror && typeof cm.CodeMirror.getValue === 'function') return cm.CodeMirror.getValue();
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent && cmContent.cmView && cmContent.cmView.view) return cmContent.cmView.view.state.doc.toString();
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.prototype.slice.call(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 6)] = codeFromCell(cell);
    });
    return code;
  }

  function collectAnswers() {
    var answers = {};
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var checked = document.querySelector('input[name="ch09-hw-' + question + '"]:checked');
      answers[question] = checked ? checked.value : '';
    });
    return answers;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch09-hw-feedback-list');
    list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5', 'q6', 'q7', 'q8', 'q9', 'q10'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row.querySelector('span');
      var item = feedback[question];
      row.className = 'ch09-hw-feedback-row';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch09-hw-feedback-list');
    list.hidden = true;
    grader.querySelectorAll('.ch09-hw-feedback-row').forEach(function (row) {
      row.className = 'ch09-hw-feedback-row';
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/homework-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var payload = text ? JSON.parse(text) : null;
        if (!payload || !response.ok || !payload.ok) {
          throw new Error((payload && (payload.message || payload.error)) || 'Submission API is unavailable.');
        }
        return payload;
      });
    });
  }

  function initHomeworkGrader() {
    var grader = document.getElementById('ch09-hw-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var actions = grader.querySelector('.ch09-hw-actions');
    var result = document.getElementById('ch09-hw-result');
    var submit = document.getElementById('ch09-hw-submit');
    var reset = document.getElementById('ch09-hw-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      submit.disabled = true;
      result.textContent = 'Submitting homework.';

      saveAttempt({
        homework_id: homeworkId,
        answers: collectAnswers(),
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          applyFeedback(grader, payload.feedback || {});
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initHomeworkGrader);
  } else {
    initHomeworkGrader();
  }
})();
</script>
