# Introduction

Data science helps organizations turn data into better decisions. In this chapter, you will see how data science combines business questions, data, statistical reasoning, programming, and communication. You will also build a technical foundation for later Python work by studying how programming languages represent instructions, values, and basic control flow.

- Data science begins with a practical question, often about a business or management decision.
- Data becomes useful when it is cleaned, organized, analyzed, and interpreted in context.
- Programming gives analysts a repeatable way to work with data and automate analysis.
- Programming languages have precise rules, execution models, and data representations that shape how code behaves.
- Data science work depends on both technical skill and domain knowledge.

The chapter has two main sections. The first introduces data science as a field, workflow, and career area. The second introduces programming foundations, including formal languages, interpreted and compiled execution, programming constructs, number systems, and character encoding.

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Data science | An interdisciplinary field that uses data, computing, statistics, and domain knowledge to answer questions and support decisions. |
| 2 | Data | Recorded facts, measurements, observations, or values that can be organized and analyzed. |
| 3 | Domain expertise | Knowledge about the context where a problem occurs, such as retail, finance, health care, operations, or marketing. |
| 4 | Business analytics | The use of data analysis to support business questions, performance tracking, and management decisions. |
| 5 | Business intelligence | Reporting and dashboard work that helps organizations monitor what has happened and how performance is changing. |
| 6 | Data analytics | Work focused on inspecting, cleaning, transforming, and interpreting data to answer practical questions. |
| 7 | Machine learning | A field that builds systems that learn patterns from data to make predictions or support automated decisions. |
| 8 | Artificial intelligence | A broad field focused on systems that perform tasks associated with human intelligence, such as prediction, language, or reasoning. |
| 9 | Descriptive analytics | Analysis that summarizes what happened using reports, dashboards, summaries, or key performance indicators. |
| 10 | Diagnostic analytics | Analysis that investigates why something happened through drill-downs, comparisons, or segmentation. |
| 11 | Predictive analytics | Analysis that estimates what is likely to happen next using forecasts, probabilities, or model scores. |
| 12 | Prescriptive analytics | Analysis that recommends actions, rules, or choices based on data and goals. |
| 13 | CRISP-DM | A common data science process model with phases for business understanding, data understanding, preparation, modeling, evaluation, and deployment. |
| 14 | Data preparation | The process of selecting, cleaning, constructing, integrating, and formatting data before analysis or modeling. |
| 15 | Modeling | The phase of a data project where analysts select techniques and build models to represent patterns in data. |
| 16 | Deployment | The phase where results, reports, models, or decision tools are put into practical use. |
| 17 | Data engineer | A role focused on data collection, storage, pipelines, cleaning, and infrastructure. |
| 18 | Data analyst | A role focused on cleaning data, exploring patterns, producing reports, and communicating findings. |
| 19 | Machine learning engineer | A role focused on building, deploying, and maintaining machine learning models. |
| 20 | Data scientist | A role that often spans the full data workflow, from problem framing through modeling and communication. |
| 21 | Programming | The process of writing precise instructions that a computer can execute. |
| 22 | Algorithm | A step-by-step procedure for solving a problem or completing a task. |
| 23 | Natural language | A language people use in everyday communication, such as English or Spanish. |
| 24 | Formal language | A language designed with precise rules for a specific purpose, such as mathematics or programming. |
| 25 | Syntax | The formal rules that define how statements must be written in a programming language. |
| 26 | Interpreter | A program that runs source code by translating and executing it, often step by step. |
| 27 | Compiler | A program that translates source code into another form, such as bytecode or machine code, before execution. |
| 28 | Bytecode | An intermediate code representation that can be run by a virtual machine. |
| 29 | Sequence | A programming construct where instructions run one after another. |
| 30 | Selection | A programming construct where a program chooses between alternative paths. |
| 31 | Iteration | Repetition in a program, usually controlled by a loop. |
| 32 | Variable | A named storage location used to hold a value in a program. |
| 33 | Data type | A classification of values that determines what operations are meaningful for those values. |
| 34 | Operator | A symbol that performs an operation on one or more operands. |
| 35 | Expression | A combination of values, variables, operators, or function calls that evaluates to a value. |
| 36 | Statement | A complete instruction that the interpreter executes to perform an action or control program flow. |
| 37 | Binary | A base-2 number system that uses only 0 and 1. |
| 38 | Hexadecimal | A base-16 number system often used for memory addresses, colors, debugging, or networking. |
| 39 | Bit | The smallest unit of data, either 0 or 1. |
| 40 | Byte | A group of 8 bits, commonly used as an addressable unit of computer storage. |
| 41 | Character encoding | A system for representing text characters as numeric values that computers can store and process. |
| 42 | ASCII | An early character encoding for English letters, digits, numbers, and symbols. |
| 43 | Unicode | A standard for representing text across many writing systems. |
| 44 | UTF-8 | A dominant Unicode encoding used across the web and modern operating systems. |
