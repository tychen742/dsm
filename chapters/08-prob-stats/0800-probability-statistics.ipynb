
<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Probability | A number between 0 and 1 describing how likely an event is. |
| 2 | Event | An outcome or set of outcomes of interest. |
| 3 | Outcome | A possible result of a random process. |
| 4 | Sample space | The collection of all possible outcomes. |
| 5 | Chance model | A simplified representation of a random process. |
| 6 | Random variable | A numeric quantity whose value depends on chance. |
| 7 | Distribution | The pattern of possible values and their probabilities or frequencies. |
| 8 | Empirical distribution | A distribution built from observed data. |
| 9 | Mean | The arithmetic average of numeric values. |
| 10 | Median | The middle value when observations are ordered. |
| 11 | Standard deviation | A measure of typical distance from the mean. |
| 12 | Variance | The squared version of standard deviation. |
| 13 | Percentile | A value below which a given percentage of observations fall. |
| 14 | Sample | A subset of observations drawn from a larger population. |
| 15 | Population | The full group or process being studied. |
| 16 | Parameter | A numeric summary of a population. |
| 17 | Statistic | A numeric summary computed from a sample. |
| 18 | Sampling variation | The natural difference among statistics from different samples. |
| 19 | Simulation | Using repeated random trials to study possible outcomes. |
| 20 | Descriptive statistics | Numeric summaries that describe a dataset. |
| 21 | Uncertainty | Limited knowledge about an outcome, parameter, or decision. |
