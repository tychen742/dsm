```python
### Question 1: Checking a Probability Model
# The support team models how long a ticket takes to resolve.
#   1. Check the model: add up the probability column, rounded to 2 decimals.
#   2. Use the complement to find the probability that a ticket is NOT
#      resolved the same day, rounded to 2 decimals.
# Print exactly:
# Total probability: 1.0
# P(not same day): 0.45
import numpy as np
import pandas as pd

tickets = pd.DataFrame({
    "outcome": ["same day", "next day", "2+ days"],
    "probability": [0.55, 0.30, 0.15],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

tickets = pd.DataFrame({
    "outcome": ["same day", "next day", "2+ days"],
    "probability": [0.55, 0.30, 0.15],
})

total = round(tickets['probability'].sum(), 2)
p_same_day = tickets.loc[tickets['outcome'] == 'same day', 'probability'].sum()
p_not_same_day = round(1 - p_same_day, 2)

print("Total probability:", total)
print("P(not same day):", p_not_same_day)
```

```python
### Question 2: Expected Revenue
# New subscribers choose one of three plans with the probabilities below.
#   1. Compute the expected monthly revenue per subscriber: the sum of
#      price times probability, rounded to 2 decimals.
#   2. Use it to estimate the monthly revenue from 2,000 new subscribers.
# Print exactly:
# Expected revenue per subscriber: 24.5
# Expected revenue from 2000 subscribers: 49000.0
import numpy as np
import pandas as pd

plans = pd.DataFrame({
    "plan": ["Basic", "Plus", "Pro"],
    "price": [10, 25, 60],
    "probability": [0.5, 0.3, 0.2],
})
subscribers = 2000
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

plans = pd.DataFrame({
    "plan": ["Basic", "Plus", "Pro"],
    "price": [10, 25, 60],
    "probability": [0.5, 0.3, 0.2],
})
subscribers = 2000

per_subscriber = round((plans['price'] * plans['probability']).sum(), 2)

print("Expected revenue per subscriber:", per_subscriber)
print("Expected revenue from", subscribers, "subscribers:", per_subscriber * subscribers)
```

```python
### Question 3: Mean or Median?
# Operations wants one number for a "typical" delivery time (in days).
#   1. Compute the mean and the median of days.
#   2. If the mean is more than 20% above the median, the data are skewed
#      to the right, so report "median" as the typical value; otherwise report "mean".
# Print exactly:
# Mean: 6.5
# Median: 4.0
# Typical value: median
import numpy as np
import pandas as pd

deliveries = pd.DataFrame({
    "days": [3, 4, 4, 3, 5, 4, 22, 3, 6, 11],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

deliveries = pd.DataFrame({
    "days": [3, 4, 4, 3, 5, 4, 22, 3, 6, 11],
})

mean_days = deliveries['days'].mean()
median_days = deliveries['days'].median()

if mean_days > 1.2 * median_days:
    typical = "median"
else:
    typical = "mean"

print("Mean:", mean_days)
print("Median:", median_days)
print("Typical value:", typical)
```

```python
### Question 4: Spread of Daily Sales
# A store manager wants to know how much daily sales vary.
#   1. Compute the 25th and 75th percentiles of sales with np.percentile().
#   2. Compute the interquartile range (IQR): 75th percentile minus 25th percentile.
#   3. Compute the standard deviation with np.std(), rounded to 1 decimal.
# Print exactly:
# 25th percentile: 1125.0
# 75th percentile: 1475.0
# IQR: 350.0
# Standard deviation: 189.7
import numpy as np
import pandas as pd

daily = pd.DataFrame({
    "sales": [1000, 1200, 1100, 1500, 1300, 1400, 1600, 1300, 1500, 1100],
})
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

daily = pd.DataFrame({
    "sales": [1000, 1200, 1100, 1500, 1300, 1400, 1600, 1300, 1500, 1100],
})

q25 = np.percentile(daily['sales'], 25)
q75 = np.percentile(daily['sales'], 75)

print("25th percentile:", q25)
print("75th percentile:", q75)
print("IQR:", q75 - q25)
print("Standard deviation:", round(np.std(daily['sales']), 1))
```

```python
### Question 5: Sample Size and Standard Error
# About 20% of customers use a coupon. A manager plans a survey to estimate that proportion.
# The standard error of a sample proportion is sqrt(p * (1 - p) / n).
#   1. Compute the standard error for a sample of n = 100 customers.
#   2. Compute it again for n = 400.
#   3. Round both to 3 decimals. How does quadrupling the sample change the SE?
# Print exactly:
# SE with n=100: 0.04
# SE with n=400: 0.02
import numpy as np
import pandas as pd

p = 0.2
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

p = 0.2

se_100 = round(np.sqrt(p * (1 - p) / 100), 3)
se_400 = round(np.sqrt(p * (1 - p) / 400), 3)

print("SE with n=100:", se_100)
print("SE with n=400:", se_400)
# Four times the sample size gives half the standard error.
```

<script>
(function () {
  var labId = 'ch08-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';

    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var cm = cell.querySelector('.CodeMirror');
    if (cm && cm.CodeMirror && typeof cm.CodeMirror.getValue === 'function') return cm.CodeMirror.getValue();
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent && cmContent.cmView && cmContent.cmView.view) return cmContent.cmView.view.state.doc.toString();
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.prototype.slice.call(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch08-lab-feedback-list');
    list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row.querySelector('span');
      var item = feedback[question];
      row.className = 'ch08-lab-feedback-row';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch08-lab-feedback-list');
    list.hidden = true;
    grader.querySelectorAll('.ch08-lab-feedback-row').forEach(function (row) {
      row.className = 'ch08-lab-feedback-row';
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var payload = text ? JSON.parse(text) : null;
        if (!payload || !response.ok || !payload.ok) {
          throw new Error((payload && (payload.message || payload.error)) || 'Submission API is unavailable.');
        }
        return payload;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch08-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var actions = grader.querySelector('.ch08-lab-actions');
    var result = document.getElementById('ch08-lab-result');
    var submit = document.getElementById('ch08-lab-submit');
    var reset = document.getElementById('ch08-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      submit.disabled = true;
      result.textContent = 'Submitting code.';

      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          applyFeedback(grader, payload.feedback || {});
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
