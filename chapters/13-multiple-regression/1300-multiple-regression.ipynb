# Multiple Regression

This chapter extends simple regression to models with several predictors. You will interpret coefficients, assess model assumptions, and evaluate prediction quality with training and testing data.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Multiple regression | Regression with more than one predictor. |
| 2 | Predictor | An input variable used in a model. |
| 3 | Target | The outcome variable being modeled. |
| 4 | Coefficient | A model parameter attached to a predictor. |
| 5 | Intercept | The predicted target when all predictors equal zero. |
| 6 | Holding constant | Interpreting one predictor while other predictors stay fixed. |
| 7 | Feature matrix | A table of predictor values used for modeling. |
| 8 | Response vector | The target values used for modeling. |
| 9 | Multicollinearity | Strong association among predictors. |
| 10 | Additivity | A model assumption that predictor effects combine by addition. |
| 11 | Linearity | A model assumption that relationships are represented by straight-line effects. |
| 12 | Residual | Observed value minus predicted value. |
| 13 | Residual diagnostic | A check based on residual patterns. |
| 14 | Training data | Data used to fit a model. |
| 15 | Test data | Data reserved to evaluate model performance. |
| 16 | Mean squared error | Average squared prediction error. |
| 17 | Root mean squared error | Square root of mean squared prediction error. |
| 18 | R-squared | The proportion of outcome variation explained by a model. |
| 19 | Overfitting | When a model fits training data too closely and generalizes poorly. |
| 20 | Interpretability | How easily a model can be explained. |
| 21 | Prediction pipeline | The steps used to prepare data, fit a model, and evaluate predictions. |
