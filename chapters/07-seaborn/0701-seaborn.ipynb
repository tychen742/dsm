## Overview

Seaborn is a high-level visualization library built on top of Matplotlib. It gives you cleaner defaults and concise plotting functions for statistical graphics, while still letting you use Matplotlib for fine-grained control.

In this notebook, we will use the following learning path:

1. A first look at Seaborn plots
2. Compare Seaborn with pandas and Matplotlib
3. The Seaborn data model: semantic mappings (`hue`, `style`, `size`) and tidy, wide, and nested data
4. Figure-level vs. axes-level functions

The plot families (distribution, relational, categorical) come in the next section, followed by faceting, multivariate views, and styling.

By convention, Seaborn is imported as `sns`:

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# %pip install seaborn
import seaborn as sns

# Set consistent defaults for all following plots
sns.set_theme(style='whitegrid', context='notebook')

tips = sns.load_dataset('tips')
````

      ↑           ↑         ↑          ↑          ↑

Let's take a look at the `tips` dataset.

````python
# df.loc[row_label, column_label]

tips.loc[:, ['total_bill', 'tip', 'size', 'day', 'time', 'sex']].head()
````

Using the Matplotlib `subplots()` function, you can place the plots side-by-side. The **`ax=` parameter** tells each plotting function which subplot panel to draw on; without it, every call would create its own separate figure. pandas `df.plot()` accepts the same `ax=` argument, which is how the comparisons below put a pandas plot and a Seaborn plot side by side. You can see that the Seaborn syntax for calling the plotting functions is familiar but the method names are different. Here we have one plot created by Matplotlib and three by Seaborn.

Style note: these plots use the global theme set earlier with `sns.set_theme(style='whitegrid', context='notebook')`.

````python
fig, ax = plt.subplots(1, 4, figsize=(16, 3))

### Matplotlib
ax[0].hist(tips['total_bill'], bins=25, alpha=0.7)
ax[0].set_title('Matplotlib hist()')

### Seaborn
sns.histplot(data=tips, x='total_bill', bins=25, ax=ax[1])
ax[1].set_title('Histogram')

### Seaborn with hue
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', ax=ax[2])
ax[2].set_title('Scatter + hue')

### Seaborn categorical plot
sns.boxplot(data=tips, x='day', y='total_bill', ax=ax[3])
ax[3].set_title('Categorical boxplot')

plt.tight_layout()
````

## Comparing Libraries

Pandas plotting is excellent for fast, direct plotting from a `DataFrame`. Seaborn becomes stronger when you need **semantic mappings** (color/style/size by variables), cleaner statistical defaults, and grouped comparisons.

Practical rules:
- Start with Pandas for quick checks.
- Use **Seaborn** for **statistical/grouped** visuals.
- Drop to Matplotlib when you need exact low-level control.

As a comparison between the three visualization libraries:

| Library | Strength | Typical call |
|---|---|---|
| pandas | Fastest plot straight from a `DataFrame` | `df.plot(kind='hist')` |
| Matplotlib | Full control over every element | `ax.hist(df['col'])` |
| Seaborn | Statistical defaults and grouping by variables | `sns.histplot(data=df, x='col')` |

### Scatter Plots: pandas vs. Seaborn

Compare the two plots in the next cell. The Seaborn version shows more information about the data:

- `df.plot(kind='scatter')`: fast DataFrame-first plotting.
- `sns.scatterplot()`: semantic mappings (`hue`, `style`, `size`) with cleaner statistical **defaults**.

````python
fig, ax = plt.subplots(1, 2, figsize=(7, 4))

# Pandas: fast DataFrame-first plotting
tips.plot(
    kind='scatter',
    x='total_bill', y='tip',
    s=tips['size']*10,          ### s, not size, for scatter plot marker size
    alpha=0.7,                  ### add some transparency for better visibility
    ax=ax[0],
    title='df.plot(kind="scatter")',
)
# Seaborn: maps more variables (color, marker, size) with about the same amount of code
sns.scatterplot(
    data=tips,
    x='total_bill', y='tip',
    hue='time', 
    style='sex', 
    size='size',
    ax=ax[1]
)
ax[1].set_title('sns.scatterplot()')
ax[1].legend(bbox_to_anchor=(1, 1))  ### move legend outside the plot
plt.tight_layout()
````

### Histograms: pandas, Matplotlib, and Seaborn

Compare histogram APIs across pandas, Matplotlib, and Seaborn:

- **pandas**: DataFrame-oriented convenience
- **Matplotlib**: lower-level direct control
- **Seaborn**: statistical defaults and cleaner styling

````python
fig, ax = plt.subplots(1, 3, figsize=(10, 3))

data = tips[['total_bill', 'tip']]
print(data.head())

### Pandas histogram
data.plot(kind='hist', title='Pandas plot(kind="hist")', ax=ax[0])

### Matplotlib histogram
ax[1].hist(data, label=['total_bill', 'tip'])
ax[1].set_title('Matplotlib ax.hist()')
ax[1].legend()

### Seaborn histogram
sns.histplot(data=data, ax=ax[2])
ax[2].set_title('Seaborn histplot()')

plt.tight_layout()
````

The histograms are different because:
- pandas and Matplotlib use 10 bins by default, whereas Seaborn chooses the number of bins from the data (NumPy's `'auto'` rule, the larger of the Sturges and Freedman-Diaconis estimates), which may produce more or fewer bins.
- Matplotlib bars are narrow because we pass a 2-column DataFrame directly to `ax.hist()`. Matplotlib draws the two columns as side-by-side bars within each bin, so each bar takes half of the bin width.
- Seaborn draws both columns on the same bins, one on top of the other, and makes the bars partly transparent. Where a `tip` bar and a `total_bill` bar fall in the same bin, the colors mix (the small gray-purple bar near $7). Because tips and bills are on very different scales, almost all tips land in the first two bins, so in practice you would plot variables with different scales separately.

## Seaborn Data Model and Semantic Mappings

Seaborn works best with **tidy (long-form) data**:
- each column is a variable
- each row is one observation

Common semantic mappings (you already saw all three in the pandas vs. Seaborn scatter plot: `hue='time'`, `style='sex'`, `size='size'`):
- `hue`: map category/value to color
- `style`: map category to marker or line style
- `size`: map numeric values to marker size

Important distinction:
- Use `size=...` to map from data
- Use `s=...` for a constant marker size


### Data Formats

Data can be structured in different ways. Understanding these formats is essential because most Python visualization and analysis libraries, including Seaborn and pandas, expect data in a specific format. The common data formats are tidy, wide, and nested. 

| Format | Best For |
|--------|----------|
| Tidy | Seaborn, statsmodels, most analysis |
| Wide | Excel, quick inspection, some ML models |
| Nested | Raw API/JSON data before processing |

In practice, data often arrives in wide or nested format and should be converted to tidy before visualization or analysis. For example, `sns.barplot(data=df, x='category', y='value')` assumes each row is one observation. If your DataFrame is wide (one column per category), Seaborn won't work as expected and you may get confusing errors or plots. So the practical point is: if a Seaborn plot looks wrong or throws an error, check whether your DataFrame is in tidy format first — and if not, use `pd.melt()` to reshape it.

### Tidy (Long) Format

In tidy data, 

- **each row is one observation**, 
- **each column is one variable**, and 
- **each cell holds one value**. 
 
This is the preferred format for analysis and visualization.

````python
import pandas as pd

tidy = pd.DataFrame({
    'student': ['Alice', 'Alice', 'Bob', 'Bob'],
    'subject': ['math', 'english', 'math', 'english'],
    'score':   [90, 85, 78, 88]
})
tidy
````

### Wide Format

In wide data, 

- each row represents one student, and 
- each subject has its own column. 
 
This format is easier for humans to read at a glance.

````python
wide = pd.DataFrame({
    'student': ['Alice', 'Bob'],
    'math':    [90, 78],
    'english': [85, 88]
})
wide
````

### Nested Format

Nested data stores values inside **dictionaries** or **lists**, common when loading from JSON or APIs. It must be flattened before use.

````python
nested = {
    'Alice': {'math': 90, 'english': 85},
    'Bob':   {'math': 78, 'english': 88}
}
nested
````

### Converting Formats

**Wide vs. tidy data errors:**
If the column you want to pass as `x=` or `hue=` does not exist because its categories are spread across several columns, the data is in wide format. Use `pd.melt()` to convert it to tidy format first.

Start with a wide table of student scores:

````python
students = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Carol', 'David', 'Eve'],
    'gender':  ['F', 'M', 'F', 'M', 'F'],
    'grade':   [10, 10, 11, 11, 12],
    'math':    [90, 78, 85, 72, 95],
    'english': [85, 88, 80, 76, 92],
    'science': [92, 80, 88, 70, 96]
})
students
````

**Wide to tidy:** `pd.melt()` keeps the `id_vars` columns and stacks the subject columns into two new columns, `subject` and `score`.

````python
students_tidy = pd.melt(students,
                        id_vars=['student', 'gender', 'grade'],
                        var_name='subject',
                        value_name='score')
students_tidy
````

**Tidy to wide:** `pivot()` reverses the process, with one row per student and one column per subject.

````python
students_wide = students_tidy.pivot(index='student',
                                    columns='subject', values='score')
students_wide
````

**Nested to tidy:** build a DataFrame from the dictionary, turn the student names into a column, then melt.

````python
nested_tidy = (
    pd.DataFrame(nested).T
    .rename_axis('student')
    .reset_index()
    .melt(id_vars='student', var_name='subject', value_name='score')
)
nested_tidy
````

Once the data is tidy, every column can be passed straight to Seaborn. Here `subject` goes on the x-axis and `gender` is mapped to color:

````python
sns.barplot(data=students_tidy, x='subject', y='score', hue='gender',
            errorbar=None)   ### hide error bars; we only have a few students
plt.title('Average Score by Subject and Gender')
plt.show()
````

## Figure vs. Axes-Level APIs

Seaborn has two common usage patterns:

- **Figure-level** functions (e.g., `displot`, `relplot`, `catplot`) manage an entire figure/grid and return grid objects (`FacetGrid`, `JointGrid`, `PairGrid`).
- **Axes-level** functions (e.g., `histplot`, `scatterplot`, `lineplot`, `boxplot`) draw on a specific Matplotlib `ax` and return an `Axes`. Because they draw onto one Matplotlib `Axes`, you can combine them with other plots in the same figure.

Rule of thumb:
- Use figure-level when **faceting** is the main goal.
- Use axes-level for **custom** subplot composition.

### API Differences
| Aspect | Figure-level | Axes-level |
|---|---|---|
| **Returns** | `FacetGrid` · `JointGrid` · `PairGrid` | `Axes` |
| **Faceting** | Built-in via `row=`, `col=`, `col_wrap=` | Not available — use `plt.subplots()` + loop |
| **Sizing** | `height=` + `aspect=` | `plt.subplots(figsize=(w, h))` |
| **Multiple panels** | **Automatic grid** from data | `fig, axes = plt.subplots()` + pass `ax=` |
| **Layering plots** | Hard | Easy: multiple functions on the same `ax` |
| **When to use** | Grouped/faceted views, small multiples | Custom layouts, overlaying, single-axes control |

Here is the same scatter plot, split by meal time, drawn both ways. The figure-level `relplot()` builds the panels from the `col=` variable; the axes-level version needs `plt.subplots()`, a loop, and `ax=`.

````python
### Figure-level: one call, panels built from the data
sns.relplot(data=tips, x='total_bill', y='tip', hue='sex',
            col='time', height=3)
plt.show()

### Axes-level: you create the panels and fill each one
fig, axes = plt.subplots(1, 2, figsize=(7, 3), sharey=True)
for ax, meal in zip(axes, ['Lunch', 'Dinner']):
    sns.scatterplot(data=tips[tips['time'] == meal],
                    x='total_bill', y='tip', hue='sex', ax=ax)
    ax.set_title(f'time = {meal}')
plt.tight_layout()
plt.show()
````

### Function Reference

The table below summarizes common figure-level functions and their typical axes-level backends.

| Figure-level | Axes-level Functions | Typical use |
|---|---|---|
| `displot()` | `histplot()`, `kdeplot()`, `ecdfplot()` | Distributions, histograms, KDE, ECDF |
| `relplot()` | `scatterplot()`, `lineplot()` | Relationships and trends |
| `catplot()` | `stripplot()`, `swarmplot()`, `boxplot()`, `violinplot()`, `boxenplot()`, `barplot()`, `countplot()`, `pointplot()` | Categorical comparisons |
| `jointplot()` | `scatterplot()`, `kdeplot()`, `histplot()`, `regplot()`, `residplot()`, `kind="hex"` *(Matplotlib hexbin)* | Bivariate plots with marginals and/or regression |
| `pairplot()` | `scatterplot()`, `kdeplot()`, `histplot()` | Pairwise relationships across variables |
