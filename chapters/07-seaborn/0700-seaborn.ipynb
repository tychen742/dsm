
<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Seaborn | A Python library for statistical data visualization. |
| 2 | Tidy data | A table structure where each variable is a column and each observation is a row. |
| 3 | Long format | A tidy layout where repeated measurements are stored in rows. |
| 4 | Wide format | A table layout where related measurements are spread across columns. |
| 5 | Plot family | A group of Seaborn functions designed for a type of statistical question. |
| 6 | Distribution plot | A chart that shows the spread and shape of one variable. |
| 7 | Categorical plot | A chart that compares values across groups. |
| 8 | Relationship plot | A chart that shows association between variables. |
| 9 | Regression plot | A chart that overlays a fitted trend on paired data. |
| 10 | Facet | A small multiple panel split by a grouping variable. |
| 11 | Hue | A color mapping used to distinguish groups. |
| 12 | Palette | A set of colors used in a chart. |
| 13 | Theme | A reusable set of style choices. |
| 14 | Estimator | A summary function such as mean used inside some Seaborn plots. |
| 15 | Confidence interval | An uncertainty band around an estimated summary or trend. |
| 16 | KDE | A smoothed estimate of a distribution shape. |
| 17 | Pair plot | A grid that compares multiple numeric variables pairwise. |
| 18 | Heatmap | A color-coded display of values in a matrix. |
| 19 | Context | A Seaborn setting that scales plot elements for notebooks, talks, or papers. |
| 20 | Legend | A guide explaining visual encodings. |
| 21 | Small multiples | A set of similar charts repeated across groups. |
