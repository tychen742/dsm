
<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Two-sample comparison | An analysis that compares outcomes from two groups. |
| 2 | A/B test | A randomized comparison of two conditions or versions. |
| 3 | Treatment group | The group receiving the intervention or new condition. |
| 4 | Control group | The comparison group receiving the baseline condition. |
| 5 | Outcome | The measured result used to compare groups. |
| 6 | Difference in means | The mean of one group minus the mean of another. |
| 7 | Difference in proportions | The proportion of one group minus the proportion of another. |
| 8 | Random sampling | Selecting observations from a population by chance. |
| 9 | Random assignment | Assigning units to conditions by chance. |
| 10 | Permutation | A rearrangement used to simulate group labels under a null model. |
| 11 | Null model | A model in which group labels have no effect on outcomes. |
| 12 | Test statistic | A summary used to measure the group difference. |
| 13 | P-value | A probability measuring how unusual the observed statistic is under the null model. |
| 14 | Causality | A claim that changing one factor changes an outcome. |
| 15 | Confounding | Mixing of effects that makes causal interpretation difficult. |
| 16 | Blinding | Hiding treatment status to reduce bias. |
| 17 | Placebo | An inactive treatment used as a comparison. |
| 18 | Sample size | The number of observations in a study or group. |
| 19 | Effect size | The magnitude of a difference in practical units. |
| 20 | Statistical power | The chance that a test detects a real effect. |
| 21 | Business experiment | A structured test used to support a management decision. |
