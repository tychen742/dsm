# Matplotlib

This chapter moves from quick pandas plots to explicit Matplotlib control. You will learn the figure-and-axes model, common plotting commands, and the options that make charts reproducible and easier to adapt.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Matplotlib | A Python library for creating static, animated, and interactive visualizations. |
| 2 | Figure | The full canvas that holds one or more plots. |
| 3 | Axes | The plotting area where data, ticks, labels, and marks appear. |
| 4 | Axis | A numeric or categorical scale within an Axes. |
| 5 | Artist | Any drawable element in a Matplotlib figure. |
| 6 | Object-oriented interface | A Matplotlib style that stores figures and axes in variables. |
| 7 | Subplot | One Axes placed inside a larger Figure layout. |
| 8 | Line plot | A plot connecting ordered observations with line segments. |
| 9 | Scatter plot | A plot showing paired numeric values as points. |
| 10 | Bar plot | A plot comparing values across categories with rectangles. |
| 11 | Histogram | A plot showing the distribution of numeric values across bins. |
| 12 | Bin | An interval used to group numeric values in a histogram. |
| 13 | Marker | A symbol used to draw a point. |
| 14 | Color | A visual encoding that can distinguish groups or emphasize marks. |
| 15 | Line style | The pattern used to draw a line. |
| 16 | Legend | A guide explaining visual encodings. |
| 17 | Title | Text naming the main message or subject of a plot. |
| 18 | Axis label | Text explaining what an axis measures. |
| 19 | Tick | A labeled reference point on an axis. |
| 20 | Figure size | The width and height of the plot canvas. |
| 21 | Style sheet | A reusable set of Matplotlib appearance settings. |
