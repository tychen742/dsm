# NumPy Arrays

<h2>Overview</h2>

NumPy gives Python a fast, compact way to represent and compute with numerical collections. In this chapter, you will move from ordinary Python lists to arrays that can store business quantities, support whole-column calculations, and make simulation practical.

- NumPy arrays store numerical data in structured, efficient collections.
- Array shape, size, dimension, and data type describe how values are organized.
- Indexing and slicing select specific values, rows, columns, or ranges.
- Vectorized operations and universal functions apply calculations to whole arrays without manual loops.
- Random number generation supports simulation and uncertainty modeling in management decisions.

The chapter has three sections. The first introduces array creation, attributes, indexing, and reshaping. The second shows how vectorized computation, universal functions, aggregation, broadcasting, masking, and sorting make array work efficient. The third introduces random number generation and simulation.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | NumPy | A Python library for fast numerical computing with arrays. |
| 2 | Array | A structured collection of values that NumPy can process efficiently. |
| 3 | `ndarray` | NumPy's N-dimensional array object, used to store homogeneous values. |
| 4 | Dimension | One direction or level of an array, such as rows, columns, or deeper axes. |
| 5 | Axis | A numbered direction along which NumPy organizes or reduces array values. |
| 6 | Shape | A tuple that reports the length of an array along each axis. |
| 7 | Size | The total number of elements in an array. |
| 8 | `dtype` | The NumPy data type used to store the values in an array. |
| 9 | Homogeneous array | An array whose elements all share one data type. |
| 10 | Scalar | A single numerical value. |
| 11 | Vector | A one-dimensional array of values. |
| 12 | Matrix | A two-dimensional array organized by rows and columns. |
| 13 | Tensor | An array with three or more dimensions, or a general term for multidimensional arrays. |
| 14 | Indexing | Selecting an array element by position. |
| 15 | Slicing | Selecting a range of elements from an array. |
| 16 | Boolean mask | An array of `True` and `False` values used to filter or modify another array. |
| 17 | Fancy indexing | Selecting array values with arrays or lists of integer positions. |
| 18 | Reshaping | Changing an array's shape without changing its underlying values. |
| 19 | Transpose | An operation that swaps array axes, often turning rows into columns. |
| 20 | Broadcasting | NumPy's rule system for combining arrays with compatible but different shapes. |
| 21 | Vectorized operation | An operation applied to whole arrays at once instead of through an explicit Python loop. |
| 22 | Universal function | A NumPy function, or ufunc, that applies fast element-wise operations to arrays. |
| 23 | Element-wise operation | A calculation performed separately on corresponding elements of an array. |
| 24 | Aggregation | A calculation that summarizes many values into fewer values, often one value. |
| 25 | Axis parameter | A function argument that controls which array direction an operation works across. |
| 26 | Sorting | Arranging array values in order. |
| 27 | View | An array object that refers to data owned by another array. |
| 28 | Copy | A separate array with its own data. |
| 29 | Random number generator | A tool that produces pseudorandom values for sampling and simulation. |
| 30 | Seed | A starting value that makes a pseudorandom sequence reproducible. |
| 31 | `default_rng()` | NumPy's recommended function for creating a modern random-number generator. |
| 32 | Uniform distribution | A distribution where values in a range are equally likely. |
| 33 | Normal distribution | A bell-shaped distribution described by a mean and standard deviation. |
| 34 | Sampling | Selecting values randomly from a population, list, or distribution. |
| 35 | Simulation | Repeatedly using random values to model uncertainty or possible outcomes. |
