```python
### Solution
import numpy as np
import pandas as pd

spend = pd.Series([15, 22, 61, 30, 75, 58, 90])
centers = [20, 70]

assignments = []
for value in spend:
    if abs(value - centers[0]) <= abs(value - centers[1]):
        assignments.append(0)
    else:
        assignments.append(1)

print("Assignments:", assignments)
print("Sizes:", [assignments.count(0), assignments.count(1)])
```

```python
### 7. Inertia by Hand
# Inertia is the sum of squared distances from each point to its own cluster's center.
#   1. Compute the center of each cluster (the mean of its points).
#   2. Compute the inertia.
# Print exactly:
# Centers: [[1.5, 1.5], [9.0, 8.0]]
# Inertia: 5.0
import numpy as np
import pandas as pd

points = pd.DataFrame({
    "x": [1, 2, 8, 10, 9],
    "y": [2, 1, 9, 7, 8],
    "cluster": [0, 0, 1, 1, 1],
})

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd

points = pd.DataFrame({
    "x": [1, 2, 8, 10, 9],
    "y": [2, 1, 9, 7, 8],
    "cluster": [0, 0, 1, 1, 1],
})

centers = points.groupby('cluster')[['x', 'y']].mean()
inertia = 0
for x, y, cluster in zip(points['x'], points['y'], points['cluster']):
    inertia = inertia + (x - centers.loc[cluster, 'x']) ** 2 + (y - centers.loc[cluster, 'y']) ** 2

print("Centers:", centers.values.tolist())
print("Inertia:", inertia)
```

```python
### Solution
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans

stores = pd.DataFrame({
    "traffic_su": [-1.2, -1.0, -1.3, -0.9, 0.1, 0.3, 0.0, 0.2, 1.3, 1.1, 1.4, 1.2],
    "basket_su": [1.1, 0.9, 1.2, 1.0, -1.0, -1.2, -0.9, -1.1, 0.2, 0.0, 0.3, 0.1],
})

labels = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(stores)
print("Cluster sizes:", sorted(pd.Series(labels).value_counts().tolist()))
```

```python
### 9. How Many Store Groups?
#   1. For k = 2, 3, and 4, fit KMeans(n_clusters=k, n_init=10, random_state=0) to
#      stores and compute the silhouette score, rounded to 3 decimals.
#   2. Print each score, then the k with the highest score.
# Print exactly:
# k=2: 0.682
# k=3: 0.864
# k=4: 0.773
# Best k: 3
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

stores = pd.DataFrame({
    "traffic_su": [-1.2, -1.0, -1.3, -0.9, 0.1, 0.3, 0.0, 0.2, 1.3, 1.1, 1.4, 1.2],
    "basket_su": [1.1, 0.9, 1.2, 1.0, -1.0, -1.2, -0.9, -1.1, 0.2, 0.0, 0.3, 0.1],
})

### Your code starts here.



### Your code ends here.
```

```python
### Solution
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

stores = pd.DataFrame({
    "traffic_su": [-1.2, -1.0, -1.3, -0.9, 0.1, 0.3, 0.0, 0.2, 1.3, 1.1, 1.4, 1.2],
    "basket_su": [1.1, 0.9, 1.2, 1.0, -1.0, -1.2, -0.9, -1.1, 0.2, 0.0, 0.3, 0.1],
})

scores = {}
for k in [2, 3, 4]:
    labels = KMeans(n_clusters=k, n_init=10, random_state=0).fit_predict(stores)
    scores[k] = round(silhouette_score(stores, labels), 3)
    print("k=" + str(k) + ":", scores[k])
print("Best k:", max(scores, key=scores.get))
```

```python
### Solution
import numpy as np
import pandas as pd

segments = pd.DataFrame({
    "visits": [2, 3, 12, 14, 1, 13, 6, 5, 7],
    "spend": [120, 140, 40, 50, 100, 45, 80, 70, 90],
    "cluster": [0, 0, 1, 1, 0, 1, 2, 2, 2],
})

profile = segments.groupby('cluster')[['visits', 'spend']].mean()
top = profile['visits'].idxmax()

print("Most frequent visitors: cluster", top)
print("Their mean spend:", profile.loc[top, 'spend'])
print("Customers:", (segments['cluster'] == top).sum())
# Frequent visitors spend the least per month: possibly a segment to target with
# bundles or add-ons.
```

(function () {
  var homeworkId = 'ch15-homework';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';

    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var cm = cell.querySelector('.CodeMirror');
    if (cm && cm.CodeMirror && typeof cm.CodeMirror.getValue === 'function') return cm.CodeMirror.getValue();
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent && cmContent.cmView && cmContent.cmView.view) return cmContent.cmView.view.state.doc.toString();
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

    var cells = Array.prototype.slice.call(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 6)] = codeFromCell(cell);
    });
    return code;
  }

  function collectAnswers() {
    var answers = {};
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var checked = document.querySelector('input[name="ch15-hw-' + question + '"]:checked');
      answers[question] = checked ? checked.value : '';
    });
    return answers;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch15-hw-feedback-list');
    list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5', 'q6', 'q7', 'q8', 'q9', 'q10'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row.querySelector('span');
      var item = feedback[question];
      row.className = 'ch15-hw-feedback-row';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch15-hw-feedback-list');
    list.hidden = true;
    grader.querySelectorAll('.ch15-hw-feedback-row').forEach(function (row) {
      row.className = 'ch15-hw-feedback-row';
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

      credentials: 'same-origin',

        if (!payload || !response.ok || !payload.ok) {
          throw new Error((payload && (payload.message || payload.error)) || 'Submission API is unavailable.');
        }
        return payload;
      });
    });
  }

  function initHomeworkGrader() {
